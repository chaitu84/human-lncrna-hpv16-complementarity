# Global HPV-18 Burden and Cervical Cancer Projections (2023-2050)

This notebook generates a structured Excel spreadsheet modeling the global burden, natural history, and 2050 public health projections of high-risk HPV strains (primarily HPV-18 and HPV-16) using data from the World Health Organization (WHO) and *Nature Medicine* (2026).

### Approach
- **Worldwide Scope:** Aggregated data reflecting global impacts rather than specific nations.
- **Non-Gender Biased:** Focuses on the overarching viral dynamics, clearance rates, global mortality totals, and universal public health elimination frameworks.

```python
import pandas as pd
import os

# Create dataframes for each tab

# 1. Global Projections 2023-2050
projections_data = {
    'Metric / Indicator': [
        'Global Annual Cervical Cancer Cases',
        'Global Annual Cervical Cancer Deaths',
        'Projected Rise in Global Cases by 2050 (Stagnant Stance)',
        'Projected Rise in Global Deaths by 2050 (Stagnant Stance)'
    ],
    '2023 Baseline': ['887,000', '369,000', 'Baseline', 'Baseline'],
    '2050 Projection': ['1,080,000', '522,000', '+20.2%', '+39.3%'],
    'Primary Data Source / Citation': [
        'WHO Fact Sheets / Nature Medicine Forecast (2026)',
        'WHO Fact Sheets / Nature Medicine Forecast (2026)',
        'Nature Medicine Global Burden Forecasts (2026)',
        'Nature Medicine Global Burden Forecasts (2026)'
    ]
}
df_projections = pd.DataFrame(projections_data)

# 2. Prevalence & Clearance
prevalence_data = {
    'Epidemiological Marker': [
        'Lifetime Risk of Contracting Any HPV Strain',
        'Global High-Risk Strain Active Point Prevalence',
        'Specific HPV-18 Strain Point Prevalence',
        'Natural Immune Clearance Rate (within 1-2 years)',
        'Persistent High-Risk Infection Rate (Progression Risk)'
    ],
    'Value / Proportion': [
        '~80% to 90% of sexually active population',
        '11.0% - 12.0% of individuals globally',
        '1.3% - 1.7% of total population',
        '~90.0% of all cases',
        '~10.0% of all cases'
    ],
    'Scientific Citation / Study Reference': [
        'CDC / World Health Organization (WHO) Epidemiology Guidelines',
        'WHO Human Papillomavirus and Cancer Technical Reports',
        'NHANES Population Studies (published in JAMA Oncology)',
        'WHO Global Cervical Cancer Screening Standard Frameworks',
        'WHO Global Cervical Cancer Screening Standard Frameworks'
    ]
}
df_prevalence = pd.DataFrame(prevalence_data)

# 3. WHO 2030 Global Elimination Strategy
strategy_data = {
    'Target Operational Pillar': [
        '90% Vaccination Coverage',
        '70% Screening Coverage',
        '90% Treatment Infrastructure',
        'Long-Term Public Health Impact (Next Century)'
    ],
    'Strategic Target Specification': [
        'Fully vaccinate 90% of girls/eligible cohorts by age 15 with 9-valent vaccine',
        'Screen 70% of individuals twice in lifetime using high-performance HPV test',
        'Provide managed treatment for 90% of individuals identified with pre-cancer/cancer',
        'Avert an estimated 74 million cancer cases and prevent 62 million deaths'
    ],
    'Official Policy Source': [
        'WHO Global Strategy to Accelerate the Elimination of Cervical Cancer',
        'WHO Global Strategy to Accelerate the Elimination of Cervical Cancer',
        'WHO Global Strategy to Accelerate the Elimination of Cervical Cancer',
        'WHO Cervical Cancer Elimination Initiative Modeling Projections'
    ]
}
df_strategy = pd.DataFrame(strategy_data)

# Create the directories if they do not exist
os.makedirs('generated', exist_ok=True)

# Write dataframes to a multi-tab Excel file
output_path = 'generated/hpv18_mortality_and_projections-v2.xlsx'
with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    df_projections.to_excel(writer, sheet_name='Global Projections 2023-2050', index=False)
    df_prevalence.to_excel(writer, sheet_name='Prevalence & Clearance', index=False)
    df_strategy.to_excel(writer, sheet_name='WHO 2030 Targets', index=False)

print(f'Successfully built multi-tab spreadsheet saved at: {output_path}')
```
