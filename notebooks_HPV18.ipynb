### 2. Prevalence & Natural History Dataset

    *   Isolates the specific, narrow point-prevalence of the raw HPV-18 strain to between **1.3% and 1.7%** of the global population.
*   **Natural Clearance Kinetic Rates:** WHO Oncology Frameworks & Clinical Meta-Analyses.
    *   Documents that **~90.0%** of all detected high-risk infections successfully resolve naturally via active host immune system clearance within a 12-to-24-month window without causing cellular abnormalities.
    *   Isolates the remaining **~10.0%** of cases as chronic, persistent infections tracking a risk of mutative transition into Cervical Intraepithelial Neoplasia (CIN 2/3).
*   **Oncogenic Strain Attributability:** The Lancet Oncology Meta-Analyses and WHO Cancer Research Registries.
    *   Attributes a definitive **70.0% to 76.0%** of all global invasive cervical tumors directly to co-infections or primary profiles of strains HPV-16 and HPV-18.
